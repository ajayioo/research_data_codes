## libraries...

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix
```

## files...

```python
df = pd.read_excel("C:\\ajayioo\\Research\\my wife\\papers\\data\\predictive_analytics_data.xlsx")
```

## extracting and renaming relevant files...

```python
df = df.iloc[:, [1, 2, 10, 11, 12, 13, 14, 16]]
df.columns = [
    "School_Type", "Position", "Experience", "Intervention_Frequency",
    "Teaching_Strategy", "Reported_Challenge", "Learning_Gap", "Resource_Constraint"
]
```

## missing values...

```python
df = df.dropna(subset=["Learning_Gap", "School_Type", "Experience", "Teaching_Strategy"])
```

## converting learning gaps to binary...

```python
df["Learning_Gap_Binary"] = df["Learning_Gap"].apply(lambda x: 1 if "Yes" in str(x) else 0)
```

## encoding...

```python
le = LabelEncoder()
for col in ["School_Type", "Experience", "Intervention_Frequency", 
            "Teaching_Strategy", "Reported_Challenge", "Resource_Constraint"]:
    df[col] = le.fit_transform(df[col].astype(str))
```

## defining features and targets...

```python
X = df[["School_Type", "Experience", "Intervention_Frequency", 
        "Teaching_Strategy", "Reported_Challenge", "Resource_Constraint"]]
y = df["Learning_Gap_Binary"]
```

## Data Splitting...

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

## training decision tree model

```python
clf = DecisionTreeClassifier(random_state=42)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
```

## evaluating model...

```python
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

## predicting samples...

```python
sample_predictions = X_test.copy()
sample_predictions["Actual"] = y_test
sample_predictions["Predicted"] = y_pred
```

## resetting index...

```python
sample_predictions.reset_index(drop=True, inplace=True)
```

## Decoding categorical variables back to original text

```python
label_columns = ["School_Type", "Experience", "Intervention_Frequency", 
                 "Teaching_Strategy", "Reported_Challenge", "Resource_Constraint"]

label_encoders = {}
for col in label_columns:
    le = LabelEncoder()
    le.fit(df[col].astype(str))
    label_encoders[col] = le
    sample_predictions[col] = le.inverse_transform(sample_predictions[col])
```
